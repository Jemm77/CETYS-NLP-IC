# 9.1 Export an Iris model

Train, export, and reuse a model in the supplied console application. Read `../README.md`.
Add new code cells after each implementation prompt and before its assertions.
Preserve all supplied cells, IDs, and assertions. No EDA or preprocessing is required.


```python
from pathlib import Path
import joblib
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.base import clone
from sklearn.linear_model import LogisticRegression

ACTIVITY_DIR = Path.cwd() if (Path.cwd() / "src").is_dir() else Path.cwd().parent
MODELS_DIR = ACTIVITY_DIR / "models"
MODEL_PATH = MODELS_DIR / "iris.joblib"
iris = load_iris()
class_names = iris.target_names.tolist()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42, stratify=iris.target
)
```

## 1. Train and evaluate — 10 points

Create `model = LogisticRegression(max_iter=1000, random_state=42)` and fit it
only on `X_train, y_train`. Keep the other estimator settings at their defaults.
Store `test_predictions = model.predict(X_test)` and `test_accuracy` using
`accuracy_score`. Display the test accuracy. Use this same fitted model for
joblib export and the optional ONNX notebook; do not substitute another estimator.
LogisticRegression supports both class probabilities and ONNX conversion.
No minimum accuracy, EDA, preprocessing, or tuning is required.


```python
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)

test_predictions = model.predict(X_test)
test_accuracy = accuracy_score(y_test, test_predictions)
```

```python
from sklearn.utils.validation import check_is_fitted
assert isinstance(model, LogisticRegression)
assert model.get_params() == LogisticRegression(max_iter=1000, random_state=42).get_params()
check_is_fitted(model)
np.testing.assert_array_equal(model.classes_, [0, 1, 2])
np.testing.assert_array_equal(test_predictions, model.predict(X_test))
assert np.isclose(test_accuracy, accuracy_score(y_test, test_predictions))
assert 0 <= test_accuracy <= 1
reference_model = clone(model).fit(X_train, y_train)
np.testing.assert_array_equal(model.predict(iris.data), reference_model.predict(iris.data))
```

## 2. Export — 25 points

Use joblib to write `MODEL_PATH` (`models/iris.joblib`) with a dictionary containing
`"model"` (the fitted estimator) and `"class_names"` (the supplied species list).
Do not retrain or include test data. Upload this exact file to Google Drive, allow
the teacher to download it, and add its share link in a new markdown cell here.
The teacher checks the download manually: a GitHub copy alone is insufficient.


```python
joblib.dump({"model": model, "class_names": class_names}, MODEL_PATH)
```

```python
assert MODEL_PATH.is_file()
exported = joblib.load(MODEL_PATH)
assert exported["class_names"] == class_names
np.testing.assert_array_equal(exported["model"].classes_, model.classes_)
np.testing.assert_array_equal(exported["model"].predict(iris.data), model.predict(iris.data))
assert exported["model"].get_params(deep=True) == model.get_params(deep=True)
```
